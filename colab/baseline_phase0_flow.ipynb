Google Drive 공유 폴더에 있는 프로젝트 루트를 찾아 `scripts/train_baseline.py`를 실행하는 thin notebook입니다.

```python
from pathlib import Path
import os

try:
    from google.colab import drive  # type: ignore
    drive.mount('/content/drive')
except Exception:
    print('Not running in Colab or Drive mount skipped.')

PROJECT_ROOT_OVERRIDE = None  # 예: Path('/content/drive/MyDrive/Graduate-Project/Falling-Model-Development')
PROJECT_ROOT_CANDIDATES = [
    Path('/content/drive/MyDrive/Graduate-Project/Falling-Model-Development'),
    Path('/content/drive/MyDrive/Falling-Model-Development'),
    Path('/content/drive/MyDrive/졸업 과제/Falling-Model-Development'),
    Path('/content/drive/MyDrive/졸업 과제/Falling-Detection-Development'),
    Path.cwd(),
]

if PROJECT_ROOT_OVERRIDE is not None:
    PROJECT_ROOT = Path(PROJECT_ROOT_OVERRIDE)
else:
    PROJECT_ROOT = next((p for p in PROJECT_ROOT_CANDIDATES if (p / 'scripts' / 'train_baseline.py').exists()), None)
    if PROJECT_ROOT is None:
        raise FileNotFoundError('Project root not found. Set PROJECT_ROOT_OVERRIDE.')

os.chdir(PROJECT_ROOT)
DATA_ROOT = PROJECT_ROOT / 'dataset'
OUTPUT_ROOT = PROJECT_ROOT / 'results' / 'baselines_phase0'
print('PROJECT_ROOT =', PROJECT_ROOT)
print('DATA_ROOT    =', DATA_ROOT)
print('OUTPUT_ROOT  =', OUTPUT_ROOT)
```

```python
# 필요 시 Colab 런타임에 패키지를 설치합니다.
import sys, subprocess
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'tensorflow', 'pandas', 'scikit-learn', 'matplotlib'], check=False)
```

```python
# 실행할 실험을 선택합니다.
EXPERIMENT_ID = 'B-TCN-D'  # B-TCN-raw, B-TCN-D, B-GRU-raw, B-GRU-D
SMOKE = True

config_path = PROJECT_ROOT / 'configs' / 'experiments' / 'phase0' / f'{EXPERIMENT_ID}.json'
cmd = [sys.executable, 'scripts/train_baseline.py', '--config', str(config_path), '--project-root', str(PROJECT_ROOT)]
if SMOKE:
    cmd.append('--smoke')
print(' '.join(cmd))
subprocess.run(cmd, check=True)
```

```python
# 여러 실험 실행 후 공통 비교표와 plot을 생성합니다.
cmd = [sys.executable, 'scripts/collect_baseline_results.py', '--results-root', str(OUTPUT_ROOT)]
print(' '.join(cmd))
subprocess.run(cmd, check=True)
```
